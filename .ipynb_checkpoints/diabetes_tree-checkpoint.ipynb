## Feature engineering

```python
data_df['Glucose'] = data_df['Glucose'].replace(0, np.nan)
data_df['Glucose'] = data_df['Glucose'].fillna(data_df['Glucose'].mean())
```

```python
data_df['BMI'] = data_df['BMI'].replace(0, np.nan)
data_df['BMI'] = data_df['BMI'].fillna(data_df['BMI'].mean())
```

```python

data_df['BloodPressure'] = data_df['BloodPressure'].replace(0, np.nan)
data_df['BloodPressure'] = data_df['BloodPressure'].fillna(data_df['BloodPressure'].mean())
```

## ML

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, classification_report
```

```python
X = data_df.drop('Outcome', axis=1)
```

```python
y = data_df['Outcome']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=45)
```

```python
tree_clf = DecisionTreeClassifier(max_depth=5, class_weight='balanced',min_samples_split=15, min_samples_leaf=5, random_state=45)
   
```

```python
tree_clf.fit(X_train, y_train)
```

```python
y_pred = tree_clf.predict(X_test)
```

```python
target_names = sorted(data_df['Outcome'].unique().astype(str))
```

```python
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}\n")
print(classification_report(y_test, y_pred, target_names=target_names))
#print(classification_report(y_test, y_pred, None, target_names))
```

```python
y_probs = tree_clf.predict_proba(X_test)[:, 1]
```

```python
new_= 0.65
y_pred_p = (y_probs >= new_).astype(int)
print(classification_report(y_test, y_pred_p, target_names=target_names))
```

```python
from imblearn.over_sampling import SMOTE
```

```python
smote = SMOTE(random_state=45)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)
```

```python
tree_clf_smote = DecisionTreeClassifier(max_depth=5, min_samples_split=15, min_samples_leaf=5, random_state=45)
tree_clf_smote.fit(X_train_res, y_train_res)

```

```python
y_pred = tree_clf_smote.predict(X_test)
print(classification_report(y_test, y_pred, target_names=target_names))
```

```python
from imblearn.under_sampling import RandomUnderSampler

rus = RandomUnderSampler(random_state=45)
X_train_res, y_train_res = rus.fit_resample(X_train, y_train)
```

```python
print(classification_report(y_test, y_pred, target_names=target_names))
```

### Random forest classifier

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
forest_clf = RandomForestClassifier (
    n_estimators = 50,
    max_depth = 6,
    class_weight='balanced',
    min_samples_split = 10,
    min_samples_leaf = 8,
    random_state = 45
)
```

```python
forest_clf.fit(X_train, y_train)
```

```python
y_pred = forest_clf.predict(X_test)
```

```python
print(classification_report(y_test, y_pred, target_names=target_names))
```

## Grid search

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'n_estimators': [50, 100, 150],
    'max_depth': [4, 5, 6, 8],
    'min_samples_split': [10, 15, 20],
    'min_samples_leaf': [4, 5, 8]
}
```

```python
grid_search = GridSearchCV(
    forest_clf,                  
    param_grid, 
    cv=5,                      
    scoring='f1_macro',
    verbose=2
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best paramters:", grid_search.best_params_)
```

```python
best_model = grid_search.best_estimator_
```

```python
print(best_model)
```
